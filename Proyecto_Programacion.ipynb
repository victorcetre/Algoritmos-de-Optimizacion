> <span style="color: red;">b. ¿Cuantas posibilidades hay teniendo en cuenta todas las restricciones?</span>

Para calcular el número de posibilidades teniendo en cuenta todas las restricciones, debemos desglosar el problema en dos partes: la selección de los números y la selección de los operadores.

- Seleccion de números

Disponemos de un conjunto de 9 cifras (del 1 al 9). Al analizar la estructura del problema y el ejemplo ($4+2-6/3*1$), vemos que para intercalar 4 operadores se necesitan exactamente 5 posiciones para números.

Como no podemos repetir ningún número, para la primera posición tenemos 9 opciones, para la segunda 8, para la tercera 7, para la cuarta 6, y para la quinta 5. El cálculo se representa con la fórmula de variaciones sin repetición de 9 elementos tomados de 5 en 5:

$$V(9,5) = \frac{9!}{(9-5)!} = 9 \times 8 \times 7 \times 6 \times 5 = 15,120$$

- Seleccion de operadores

Esto es una permutación de 4 elementos:

$$P(4) = 4! = 4 \times 3 \times 2 \times 1 = 24$$
